# Настройка веб-аналитики и управление рекламными кампаниями

Комплексная работа по настройке веб-аналитики для интернет-магазина: внедрение систем аналитики, настройка целей, анализ эффективности рекламных кампаний и оптимизация рекламного бюджета на основе данных.

1. **Настройка веб-аналитики:**
   - Установка счётчиков Google Analytics и Яндекс Метрики.
   - Настройка целей (воронка: просмотр → добавление в корзину → покупка).
   - Создание сквозной отчётности для оценки эффективности каналов.

2. **Анализ текущих рекламных кампаний:**
   - Оценка эффективности форматов объявлений (CTR, CR, CPA, CPO, ДРР).
   - Выявление точек роста и проблемных кампаний.

3. **Формулирование гипотез и оптимизация:**
   - Замена неэффективных креативов.
   - Обновление аудиторных сегментов.
   - Минусовка нерелевантных запросов.
   - Оптимизация посадочных страниц.
